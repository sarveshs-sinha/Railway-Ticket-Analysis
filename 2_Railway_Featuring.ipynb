```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
'''Loading Cleaned Data'''
df_clean = pd.read_csv('Railway_Cleaned.csv')
```

```python
df_clean.head(2)
```

```python
#Here we see Date of Journey and Booking Date are in string form so convert it into datetime 

df_clean['Date of Journey'] = pd.to_datetime(df_clean['Date of Journey'])

df_clean["Booking Date"] = pd.to_datetime(df_clean["Booking Date"])
```

```python
'''Featuring Engineering'''
```

```python
'''Booking Time Lead'''
df_clean["Booking Lead Time"] = (df_clean["Date of Journey"] - df_clean["Booking Date"]).dt.days 
#It tellks how many days before journey ticket was booked 

df_clean.head()
```

```python
'''Waitlist in numbers only'''

df_clean['Waitlist Number'] = (df_clean['Waitlist Position'].astype(str).str.extract(r"(\d+)")[0])

#Now convert into numeric
df_clean['Waitlist Number'] = pd.to_numeric(df_clean['Waitlist Number'])
```

```python
'''Create Routes'''

df_clean['Route'] = df_clean['Source Station'].astype(str) + "->" + df_clean['Destination Station'].astype(str)

df_clean['Route'].head() #For checking
```

```python
'''High Demand Month'''

df_clean["Journey Month"] = (df_clean["Date of Journey"].dt.month_name()) #This will extract month only

df_clean["Journey Month"].value_counts()
```

```python
'''High Demand Day'''

df_clean["Journey Day"] = (df_clean["Date of Journey"].dt.day_name()) #This will extract day only

df_clean["Journey Day"].value_counts()
```

```python
'''High Demand Year'''

df_clean["Journey Year"] = (df_clean["Date of Journey"].dt.year) #This will extract year only

df_clean["Journey Year"].value_counts()
```

```python
'''Waitlist Category'''

df_clean['Waitlist Category'] = pd.cut(df_clean['Waitlist Number'],
                                 bins= [-1, 0, 10, 50, 100, np.inf],
                                 labels= ['No WL', 'WL 1-10', 'WL 11-50', 'WL 51-100', 'WL 100+'] )

df_clean["Waitlist Category"].value_counts()
```

```python
'''Booking Lead-Time Category'''

df_clean["Booking Lead Time Category"] = pd.cut(df_clean["Booking Lead Time"],
                                         bins= [-1, 1, 7, 30, 90, np.inf],
                                         labels= ["Last Minute", "2-7 Days", "8-30 Days", "31-90 Days", "90+ Days"] )

df_clean["Booking Lead Time Category"].value_counts()
```

```python
'''Distance Category'''

df_clean["Distance Category"] = pd.cut(df_clean["Travel Distance"],
                                 bins= [-1, 300, 800, 1500, np.inf],
                                 labels= ["Short", "Medium", "Long", "Very Long"] )

df_clean["Distance Category"].value_counts()
```

```python
'''Passenger Category'''


df_clean["Passenger Group Size"] = pd.cut(df_clean["Number of Passengers"],
                                   bins=[0, 1, 2, 4, np.inf],
                                   labels=["1 Passenger", "2 Passengers", "3-4 Passengers", "5+ Passengers"] )

df_clean["Passenger Group Size"].value_counts()
```
