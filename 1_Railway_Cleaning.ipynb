```python
'''Cleaning Null Values :-
    i.e.
        Special Considerations     9955
        and Waitlist Position      19947 '''
```

```python
#Never Clean Original Dataframe use a clone or copy version

df_clean = df.copy()
```

```python
#PNR of Train must be unique so we check it also
df_clean["PNR Number"].nunique()
```

```python
df_clean.head(3)
```

```python
#What percent of data is missing from dataset

missing = df.isnull().sum()

missing_percentage = ( missing / (len(df_clean))) * 100

pd.DataFrame({
    "Missing Values": missing,
    "Missing %": missing_percentage
})
```

```python
#Special Considerations has 9955 null values represented by NaN
#Here it means there is no special consideration for the passenger

df_clean['Special Considerations'] = df_clean['Special Considerations'].fillna('No Special Consideration')
```

```python
#Waitlist Position has 19947 null values represnted by NaN
#Here it means passenger is not waitlisted so we replace it with 0 means in clean numeric value or make another column name was waitlisted in boolean form

df_clean["Was Waitlisted"] = df_clean["Waitlist Position"].notna()
```

```python
df_clean['Waitlist Position'] = df_clean['Waitlist Position'].fillna('WL0')
```

```python
df_clean['Waitlist Position'] = df_clean['Waitlist Position'].str.replace('WL', '')

df_clean['Waitlist Position'] = df_clean['Waitlist Position'].astype(int)
```

```python
#Check updated or not
df_clean.isnull().sum()
```

```python
#Here we see Date of Journey and Booking Date are in string form so convert it into datetime 

df_clean['Date of Journey'] = pd.to_datetime(df_clean['Date of Journey'])

df_clean["Booking Date"] = pd.to_datetime(df_clean["Booking Date"])
```
